```python
import pandas as pd
```

# Sales Analysis

```python
df = pd.read_csv("Superstore.csv", encoding="latin-1")
df.head()
```

```python
df.Sales.sum()
```

```python
g = df.groupby("Region")["Sales"]

for region, region_df in g:
    print(region)
    print(region_df)
```

```python
g.sum().sort_values(ascending=True)
```

The highest number of total sales took place in the West region which is 725457.8245 and least number of total sales took place in the South

```python
g1 = df.groupby("Category")["Sales"]
g1.sum().sort_values(ascending=True)
```

So the highest number of sales occured in the "Technology" category which were 836154.0330 and least number of sales occured in the "Office Supplies" category

```python
print(df.Category.unique())

g2 = df[df["Category"] == "Furniture"].groupby("Sub-Category")["Sales"]
print(g2.sum().sort_values(ascending=True))

print("\n")

g3 = df[df["Category"] == "Office Supplies"].groupby("Sub-Category")["Sales"]
print(g3.sum().sort_values(ascending=True))

print("\n")

g4 = df[df["Category"] == "Technology"].groupby("Sub-Category")["Sales"]
print(g4.sum().sort_values(ascending=True))
```

In the,

-> Furniture Category, the highest number of total sales were made by "Chairs" which is 328449.1030

-> Office Supplies Category, the highest number of total sales were made by "Storage" which is 223843.608

-> Technology Category, the highest number of total sales were made by "Phones" which is 330007.054

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
```
